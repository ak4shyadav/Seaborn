```python
sales = { 
    'Day': ['Mon','Tue','Wed','Thu','Fri'],
    'Revenue': [1200, 1500, 900, 2000, 1800]
}
df = pd.DataFrame(sales)
print(df)
print("1. Ttoal revenue:",df['Revenue'].sum())
print("2. Average daily revenue:",df['Revenue'].mean())
print("3. Day with highest revenue:",df['Day'][df['Revenue'].idxmax()])
print("4. Days where revenue > average:",df['Day'][df['Revenue']>df['Revenue'].mean()])
df.plot(df['Day'],df['Revenue'])
```
