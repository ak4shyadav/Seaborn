```python
import matplotlib.pyplot as plt
import pandas as pd


%matplotlib inline 
#render plots inline
```

```python
df = pd.read_csv("sales.csv", encoding="latin-1")
df.head()
```

TASK1

```python
monthly_sales = df.groupby("MONTH_ID", as_index=False)["SALES"].sum()

plt.plot(monthly_sales["MONTH_ID"], monthly_sales["SALES"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

TASK2

```python
plt.scatter(df["QTR_ID"], df["QUANTITYORDERED"])
plt.title("Relationship Between QTR_ID and QUANTITYORDERED")
plt.xlabel("Quarter ID")
plt.ylabel("Quantity Ordered")
plt.show()
```

TASK3

```python
categories = ["Furniture", "Technology", "Office Supplies"]
category_sales = [25000, 40000, 30000]

plt.figure()
plt.bar(categories, category_sales)
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()

plt.figure()
plt.barh(categories, category_sales)
plt.title("Sales by Category - Horizontal Bar Chart")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.show()
```

TASK4

```python
sales_2023 = [22000, 35000, 28000]
sales_2024 = [26000, 42000, 32000]
sales_2025 = [30000, 48000, 36000]

x = [0, 1, 2]
width = 0.25

plt.figure()

plt.bar(
    [i - width for i in x],
    sales_2023,
    width=width,
    label="2023"
)

plt.bar(
    x,
    sales_2024,
    width=width,
    label="2024"
)

plt.bar(
    [i + width for i in x],
    sales_2025,
    width=width,
    label="2025"
)

plt.xticks(x, categories)

plt.title("Sales Comparison by Year")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

TASK5

```python
online_sales = [15000, 25000, 18000]
store_sales = [10000, 15000, 12000]

plt.figure()

plt.bar(categories, online_sales, label="Online Sales")
plt.bar(
    categories,
    store_sales,
    bottom=online_sales,
    label="Store Sales"
)

plt.title("Online and Store Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.legend()

plt.show()
```
