```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Utube.csv')
df.head()
```

TASK1


```python
sns.relplot(x='Total Videos', y='Avg Video Length (min)', hue='Holographic Content Rating', data=df)
plt.show()

sns.relplot(x='Total Videos', y='Avg Video Length (min)', hue='Holographic Content Rating', data=df, kind = 'scatter')
plt.show()

```

TASK2

```python
sns.lineplot(x='Quantum Computing Topics', y='Avg Video Length (min)', hue='Holographic Content Rating', data=df, style = 'scatter')
plt.show()
```
