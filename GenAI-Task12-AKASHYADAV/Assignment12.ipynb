```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Utube.csv')
df.head()
```

TASK1


```python
sns.relplot(x='Total Videos', y='Avg Video Length (min)', hue='Holographic Content Rating', data=df)
plt.show()

sns.relplot(x='Total Videos', y='Avg Video Length (min)', hue='Holographic Content Rating', data=df, kind = 'scatter')
plt.show()

```

TASK2

```python
tips = sns.load_dataset('tips')
tips.head()
sns.lineplot(y='total_bill', x='day', hue='sex', data=tips, style = 'smoker')
plt.show()
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=tips)
plt.show()

sns.relplot(x="day", y="total_bill", hue="sex", size="size", data=tips, col="smoker", col_wrap=2)
plt.show()
```

TASK3

```python
sns.histplot(tips['total_bill'], bins = 20)
plt.show()
sns.kdeplot(tips['total_bill'])
plt.show()
sns.rugplot(tips['total_bill'], color = 'red', height = 0.5)
plt.show()
sns.histplot(tips['total_bill'], bins = 20, kde = True)
plt.show()
```

TASK4

```python
sns.histplot(data=tips, x='total_bill', y='tip')
plt.show()
sns.kdeplot(data=tips, x='total_bill', y='tip', fill=True)
plt.show()
```

TASK5

```python
tips.head()
sns.pairplot(tips, hue='sex', palette='Set1')
plt.show()

sns.heatmap(tips.corr(), annot=True, cmap='coolwarm')
plt.show()
```

TASK6

```python
sns.barplot(x='day', y='total_bill', hue='sex', data=tips)
plt.show()

sns.boxplot(x='day', y='total_bill', hue='sex', data=tips)
plt.show()

sns.violinplot(x='day', y='total_bill', hue='sex', data=tips)
plt.show()

sns.countplot(x='day', hue='sex', data=tips)
plt.show()
```
